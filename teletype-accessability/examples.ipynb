### 1. Цветовые схемы и двойное кодирование (Проблема дальтонизма)
Этот скрипт генерирует два графика для сравнения. Первый использует жесткую красно-зеленую палитру, которая сольется для человека с дейтеранопией. Второй использует безопасную палитру и двойное кодирование (цвет + форма маркера).

```python
import plotly.express as px

# Используем датасет iris: 3 вида цветков, кластеры хорошо разделены визуально
df = px.data.iris()

# --- ГРАФИК 1: Недоступный дизайн (Проблема красного и зеленого) ---
fig_bad = px.scatter(
    df, 
    x="sepal_width", 
    y="sepal_length", 
    color="species",
    color_discrete_map={
        "setosa": "#FF0000",      # Чистый красный
        "versicolor": "#00FF00",  # Чистый зеленый
        "virginica": "#888800"    # Оливковый
    }
    
)

# Жестко задаем крупный размер (size=16) и убираем обводку для чистоты эксперимента
fig_bad.update_traces(marker=dict(size=16, line=dict(width=0)))
fig_bad.show()

# =====================================================================

# --- ГРАФИК 2: Инженерное решение (Safe-палитра + Двойное кодирование) ---
fig_good = px.scatter(
    df, 
    x="sepal_width", 
    y="sepal_length", 
    color="species",
    color_discrete_sequence=px.colors.qualitative.Safe, # Встроенная безопасная палитра
    symbol="species" # ДВОЙНОЕ КОДИРОВАНИЕ: Круг, Квадрат, Ромб
)

# Оставляем крупный размер (size=16), но добавляем контрастную темную обводку
fig_good.update_traces(marker=dict(size=16, line=dict(width=1.5, color='DarkSlateGrey')))
fig_good.show()
```

### 2. Совместимость со скринридерами (Отказ от эксклюзивного Hover)
Скринридеры и пользователи с клавиатуры не могут навести курсор мыши, чтобы прочитать Tooltip. Критические данные должны быть выведены прямо на график.

```python
import plotly.express as px

df_top = px.data.gapminder().query("year == 2007").nlargest(5, 'pop')

# --- ВАРИАНТ 1: Данные скрыты в Tooltip ---
fig_hover_only = px.bar(
    df_top, x='country', y='pop'
)
fig_hover_only.show()

# --- ВАРИАНТ 2: Прямой вывод текста ---
fig_text_labels = px.bar(
    df_top, x='country', y='pop',
    text_auto='.2s', # Вывод форматированных значений прямо на столбцы
)
# Настройка контрастности текста внутри столбцов
fig_text_labels.update_traces(
    textfont_size=14, textangle=0, textposition="outside", cliponaxis=False
)
fig_text_labels.show()
```
