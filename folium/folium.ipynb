#### Запись данных

```python
with open(filename, mode='w', newline = '', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'category', 'lat', 'lon'])
    writer.writerows(data)
    
print(f"{filename} created")
```

#### Чтение данных из CSV файла

```python
def read_locations_to_dict(file_path):
    locations_dict = {}

    with open(file_path, mode='r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        
        for row in reader:
            cat = row['category']
            if cat not in locations_dict:
                locations_dict[cat] = []
            locations_dict[cat].append({
                'name': row['name'], 
                'lat': float(row['lat']), 
                'lon': float(row['lon'])
            })

        return locations_dict
    
my_data = read_locations_to_dict(filename)
print(my_data)
```

```python
import folium
from folium.plugins import MarkerCluster
import os
import base64
import pathlib

m = folium.Map(location=[52.03, 113.50], zoom_start=13)
marker_cluster = MarkerCluster().add_to(m)

base_path = pathlib.Path().resolve()

def get_image_base64(path):
    with open(path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

park_b64 = get_image_base64(base_path / 'icons' / 'park.png')
swim_b64 = get_image_base64(base_path / 'icons' / 'swim.png')
rest_b64 = get_image_base64(base_path / 'icons' / 'restaurant.png')
shop_b64 = get_image_base64(base_path / 'icons' / 'shop.png')

for category, places in my_data.items():
    icon_name = icon_files.get(category, 'default.png')
    full_path = base_path / 'icons' / icon_name
    
    custom_icon = folium.CustomIcon(
        icon_image=str(full_path), 
        icon_size=(30, 30),
        icon_anchor=(15, 30)
    )
    
    for place in places:
        tooltip_text = f"<b>{place['name']}</b><br>Категория: {category}"
        folium.Marker(
            location=[place['lat'], place['lon']],
            tooltip=tooltip_text,
            icon=custom_icon
        ).add_to(marker_cluster)

legend_html = f'''
     <div style="position: fixed; bottom: 50px; left: 50px; width: 150px; height: 160px; 
     border:2px solid grey; z-index:9999; font-size:14px; background-color:white; 
     opacity: 0.9; padding: 10px;">
     <b>Легенда</b><br>
     <img src="data:image/png;base64,{park_b64}" width="20px"> Парки<br>
     <img src="data:image/png;base64,{swim_b64}" width="20px"> Купание<br>
     <img src="data:image/png;base64,{rest_b64}" width="20px"> Еда<br>
     <img src="data:image/png;base64,{shop_b64}" width="20px"> ТЦ
     </div>
     '''
m.get_root().html.add_child(folium.Element(legend_html))

m
```
